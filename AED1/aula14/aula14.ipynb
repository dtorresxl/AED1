```python
def regiao(estado):
    n = ["AC", "AP", "AM", "PA", "RO", "RR", "TO"]
    no = ["AL", "BA", "CE", "MA", "PB", "PE", "PI", "RN", "SE"]
    co = ["DF", "GO", "MT", "MS"]
    s = ["ES", "MG", "RJ", "SP"]
    sul = ["PR", "SC", "RS"]

    estado = estado.upper()

    if estado in n:
        return "N"
    if estado in no:
        return "NO"
    if estado in co:
        return "CO"
    if estado in s:
        return "S"
    if estado in sul:
        return "SE"

    return "Inválido"  # ou levante erro, dependendo do seu caso
```

```python
def entrega(peso, estado, presente=False):
    """
    CLACULA O VALOR DO FRETE BASEADO NO PESO, NO ESTADO E SE É PRESENTE OU NÃO.
    OENTRADA:
    PESO = FLOAT EM GRAMAS
    ESTADO = SIGLAS DO ESTADO
    PRESENTE =  SE É PRESENTE OU NÃO
    """
    if peso <= 500:
        vp = 15
    elif peso <= 2000:
        vp = 43
    else:
        vp = 103

    r = regiao(estado)

    if r == 'N' or r == 'NE': 
        ve = 50
    elif r == 'CO' or r == 'S':
        ve = 25
    else:
        ve = 0

    vpr = 10 if presente else 0

    return vp + ve + vpr
```

```python
peso = float(input("Peso do produto: "))
estado = input("Será enviado para qual estado? ").strip()
presente = input("É para presente?: <S/N>: ").strip()

eh_presente = presente in "Ss"

valor_frete = entrega(peso, estado, eh_presente)

print("Valor do frete:", valor_frete)
```

```python
import random

nome = input("Digite seu nome: ")

v_jogador = 0
v_script = 0
empates = 0

while True:
    parar = input("Jogar? (ou digite 'sair' para parar): ")

    if parar == "sair":
        break

    jogador = input("Sua jogada (pedra/papel/tesoura): ")
    pc = random.choice(["pedra", "papel", "tesoura"])

    print("Você:", jogador, "| Script:", pc)

    if jogador == pc:
        empates += 1
    elif (jogador == "pedra" and pc == "tesoura") or (jogador == "papel" and pc == "pedra") or (jogador == "tesoura" and pc == "papel"):
        v_jogador += 1
    else:
        v_script += 1

print("\n--- Resultado Final ---")
print("Vitórias do jogador:", v_jogador)
print("Empates:", empates)
print("Derrotas do jogador:", v_script)

if v_jogador > v_script:
    vencedor = nome
elif v_script > v_jogador:
    vencedor = "script"
else:
    vencedor = "empate"

print("Vencedor:", vencedor)
```
