```python
def numeros(numero_str):
    soma = 0
    mult = 1
    
    for digito in numero_str:
        digito = int(digito)
        soma += digito
        mult *= digito
    
    return soma, mult
```

```python
input_str = input("Digite um número:")
soma, mult = numeros(input_str)
print(f"Soma: {soma}, Multiplicação: {mult}")
```
