```python
def calcular_imc(p, a):
    imc = p / (a * a)
    return imc
```

```python
def classificar_imc(imc):
    if imc < 18.5:
        return "Abaixo do peso"
    elif imc < 24.9:
        return "Peso normal"
    elif imc < 29.9:
        return "Sobrepeso"
    else :
        return "Obesidade"
   
```

```python
p = float(input("Peso (kg): "))
a = float(input("Altura (m): "))
imc = calcular_imc(p, a)
print(f"IMC: {imc:.2f} - {classificar_imc(imc)}")
```

```python
def eh_primo(n):
    if n <= 1:
        return False
    else:
        for i in range(2, n):
            if ( n % i ) == 0:
                return False
        return True
```

```python
n = int(input("Digite um numero: "))
if eh_primo(n):
    print(f"O numero {n} é Primo")
else:
    print(f"Onumero {n} não é primo")
```
